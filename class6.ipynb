# 練習：二分搜尋 Binary Search 遞迴版

前面如果已經學過 binary search，可以用 recursion 再寫一次。

前提：

**資料必須已經排序好**

````python
def binary_search_recursive(arr, low, high, target):
    if low > high:
        return -1

    mid = (low + high) // 2

    if(arr[mid] == target):
        return mid
    elif arr[mid] > target:
        return binary_search_recursive(arr, low, mid - 1, target)
    else:
        return binary_search_recursive(arr, mid + 1, high, target)

arr = [1, 3, 5, 7, 9, 11, 13]

print(binary_search_recursive(arr, 0, len(arr) - 1, 7)) # 3
print(binary_search_recursive(arr, 0, len(arr) - 1, 8)) # -1
````

# 課堂小挑戰：河內塔 Tower of Hanoi
http://wisemath.com.tw/ele/Hanoi_Tower.html

河內塔是經典遞迴題。

規則：

1. 有三根柱子 A、B、C
2. 一次只能移動一個盤子
3. 大盤子不能放在小盤子上面
4. 目標是把所有盤子從 A 移到 C

````python
def hanoi(n, start, helper, end):
  # base case
  if n == 1:
    print('把盤子從', start, '移到', end)
    return

  # recusive case
  hanoi(n-1, start, end, helper) # 藉由 C 把 A 移動到 B
  print('把盤子從', start, '移到', end)
  hanoi(n-1, helper, start, end) # 藉由 A 把 B 移動到 C

hanoi(3, 'A', 'B', 'C')

# A -> start
# B -> helper
# c -> end
````

# 1. 什麼是搜尋？

以前學過的 Search，通常是在一組資料中尋找某個值。

例如：

```python
numbers = [8, 3, 5, 2, 9]
```

尋找數字 `5` 是否存在。

但是 DFS 和 BFS 搜尋的不是單純的一排資料，而是：

- 一張地圖可以走到哪些位置
- 一個人與哪些人有連結
- 一個節點可以到達哪些節點
- 一座迷宮是否有出口
- 有幾塊互相連在一起的區域

這些問題的共同點是：

> 從一個起點開始，沿著可以前往的方向，不斷找下去。

# 3. 為什麼需要 visited？

從 A 走到 B 後，B 又可以走回 A。

```text
A → B → A → B → A → ...
```

如果沒有記錄已經走過的位置，程式可能會一直重複搜尋。

因此需要建立：

```python
visited = set()
```

當走到某個節點時，就把它加入 `visited`：

```python
visited.add("A")
```

下次準備走到 A 前，先判斷：

```python
if "A" not in visited:
```

這是 DFS 和 BFS 最重要的觀念之一。

````python
visited = set()

visited.add("A")
visited.add("B")

print("目前走過：", visited)
print("A 是否走過：", "A" in visited)
print("C 是否走過：", "C" in visited)
````

````python
# DFS 深度優先搜尋：選一條路一直走到底，不能再走時才退回上一個位置

'''
可以想像走迷宮時，規定自己：
1. 選一條路
2. 一直向前走
3. 遇到死路就退回
4. 改走另一條路

DFS 常用兩種方式實作：
1. Recursion
2. Stack
'''
````

DFS 的遞迴想法

DFS 的遞迴流程：

```text
拜訪目前節點
→ 標記為走過
→ 找出所有相鄰節點
→ 對尚未走過的相鄰節點繼續 DFS
```

程式骨架：

```python
def dfs(node):
    visited.add(node)

    for next_node in graph[node]:
        if next_node not in visited:
            dfs(next_node)
```

注意：

- `dfs(node)` 表示搜尋目前節點
- `dfs(next_node)` 表示進入下一層
- 沒有其他可走的節點時，函式會自動返回上一層

````python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A"],
    "D": ["B"],
    "E": ["B"]
}

visited = set()

def dfs(node):
    visited.add(node)
    print("走到：", node)

    for next_node in graph[node]:
        if next_node not in visited:
            dfs(next_node)

dfs("A")
````

```text
│   │
C   E
```

# DFS Step by Step

假設從 A 開始，且每次按照清單順序選擇下一個節點。

```text
A 的鄰居：B、C
B 的鄰居：A、D、E
```

流程如下：

```text
dfs(A)
├─ 標記 A
├─ 前往 B
│  ├─ 標記 B
│  ├─ A 已走過，跳過
│  ├─ 前往 D
│  │  ├─ 標記 D
│  │  └─ B 已走過，返回
│  ├─ 前往 E
│  │  ├─ 標記 E
│  │  └─ B 已走過，返回
│  └─ 返回 A
├─ 前往 C
│  ├─ 標記 C
│  └─ A 已走過，返回
└─ 搜尋結束
```

可能的走訪順序：

```text
A → B → D → E → C
```

DFS 的順序會受到鄰居排列順序影響，但只要所有可到達節點都有拜訪到，就算正確。

# 使用 Stack 實作 DFS

Recursion 的背後其實也使用了呼叫堆疊。

因此 DFS 也可以直接使用 Stack。

流程：

1. 起點放入 Stack
2. Stack 不為空時，取出最上面的節點
3. 如果沒有走過，就標記並處理
4. 把相鄰節點放進 Stack

程式骨架：

```python
stack = [start]

while stack:
    node = stack.pop()
```

因為 Stack 是後進先出，所以最後放入的節點會最先被取出。

````python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A"],
    "D": ["B"],
    "E": ["B"]
}
````

````python
graph = {
    1: [2, 3],
    2: [1, 4, 5],
    3: [1, 6],
    4: [2],
    5: [2],
    6: [3]
}

stack = [1]
visited = set()

while stack:
    node = stack.pop()
    if node not in visited:
        visited.add(node)
        print(node, end=' ')
        for next_node in reversed(graph[node]): # reversed 讓結果比較接近遞迴 DFS 的順序
            stack.append(next_node)
````

# 為什麼 Stack 版本常使用 reversed？

假設 A 的鄰居是：

```python
["B", "C"]
```

依序放進 Stack：

```text
先放 B，再放 C
```

Stack 最上面會是 C，因此取出順序變成：

```text
C → B
```

若希望先走 B，可以反過來放：

```python
for next_node in reversed(graph[node]):
```

這不是 DFS 一定要做的規則，只是為了讓走訪順序比較容易預測。
