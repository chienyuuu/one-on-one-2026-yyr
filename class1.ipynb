```python
# 建立字典
dict1 = {'John': 30, 'Eddy': 21, 'Amy': 18} # string, integer, float 都可以當 key，只是大部分的鍵都是 string
dict2 = dict([['John', 30], ['Eddy', 21], ['Amy', 18]])
dict3 = dict(John = 30, Eddy = 21, Amy = 18) # 此種方法 key 不能用數值


print(dict1['Eddy'])
print(dict2['Eddy'])
print(dict3['Eddy'])
```

```python
# 鍵重複

'''
用「鍵」作為索引來取「值」，值可以重複
當鍵重複時，後面會覆蓋掉前面，只有最後的有效
'''

dict1 = {'John': 30, 'Eddy': 21, 'Amy': 18, 'John': 40}

print(dict1['John'])
```

```python
# 鍵不存在

dict1 = {'John': 30, 'Eddy': 21, 'Amy': 18}

print(dict1['Bob']) # wrong
print(dict1[0]) # wrong
```

```python
# 用 get()，當鍵不存在時不會跳錯誤

# get(鍵, 預設值)
# 鍵存在 -> 回傳值
# 鍵不存在 且 沒有預設值 -> 回傳 None
# 鍵不存在 但 有預設值 -> 回傳 預設值

dict1 = {'John': 30, 'Eddy': 21, 'Amy': 18}

print(dict1['Eddy'])
print(dict1.get('Eddy'))
print(dict1.get('Eddy', 15))

print(dict1.get('Bob'))
print(dict1.get('Bob', 15))
```

```python
# 以鍵(key)取得元素組合

# keys()：可取得所有鍵，資料型態是 dict_keys

dict1 = {'John': 30, 'Eddy': 21, 'Amy': 18}
key1 = dict1.keys()
print(key1)

# 用 for 迴圈印出所有鍵
for i in key1:
  print(i, end=' ')

print(key1[0]) # wrong
# dict_keys 看起來像串列，但其實不是，如果要取得元素值，要先轉成 list
# dict_keys -> list
list1 = list(key1)
print(list1[0])

print(type(list1))
print(type(dict1))
print(type('123'))
print(type(123))
print(type(key1))
```

```python
# 以值(value)取得元素組合

# values()：可取得所有值，資料型態是 dict_values
dict1 = {'John':30, 'Eddy':21, 'Mary':18}
value1 = dict1.values()
print(value1)

# 用 for 迴圈印出所有值
for i in value1:
  print(i, end=' ')

print(value1[0]) # wrong

# dict_values -> list
list2 = list(value1)
print(list2[1])
```

```python
# items()：以「鍵值對」取得元素組合，資料型態是 dict_items

dict1 = {'John':30, 'Eddy':21, 'Mary':18}
item1 = dict1.items()
print(item1)

for key, value in item1:
  print(key, value)

# 資料型態轉成 list 是一個類似二維串列
list1 = list(item1)
print(list1[0])
print(list1[0][1])
```

```python
# setdefault(鍵, 預設值)
# 和 get() 類似，若 key 不存在，就用括號內的參數「建立新元素」

# 鍵存在 -> 回傳值
# 鍵不存在 且 沒有預設值 -> 回傳 None
# 鍵不存在 但 有預設值 -> 回傳 預設值

dict1 = {'John':30, 'Eddy':21, 'Mary':18}
print(dict1.setdefault('John'))
print(dict1.setdefault('John', 40))
print(dict1)

dict1 = {'John':30, 'Eddy':21, 'Mary':18}
print(dict1.setdefault('Bob'))
print(dict1)

dict1 = {'John':30, 'Eddy':21, 'Mary':18}
print(dict1.setdefault('Bob', 15))
print(dict1)
```

```python
# range()


# 一個參數
# 數列變數 = range(整數值)
# 產生 0 至 (整數值 - 1) 的串列

list1 = range(5) # 0, 1, 2, 3, 4
print(list(list1))
```

```python
# 兩個參數
# 數列變數 = range(起始值, 終止值)
# 產生 (起始值) 至 (終止值 - 1) 的串列

list2 = range(3, 8)
print(list(list2))

list3 = range(-3, 6)
print(list(list3))

'''
若起始值 >= 終止值，則產生空串列！
'''
list4 = range(8, 6)
print(list(list4))
```

```python
# 三個參數
# 數列變數 = range(起始值, 終止值, 間隔值)
# 產生 (起始值) 開始，每次遞增 (間隔值)，直到 (終止值 - 1) 的串列

list5 = range(3, 8, 2)
print(list(list5))

list6 = range(3, 7, 2)
print(list(list6))

list7 = range(5, -6, -1)
print(list(list7))
```
