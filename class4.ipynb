````python
# stack 堆疊

'''
可以想像成疊盤子：最後放上去的盤子，會最先被拿走。

後進先出 LIFO (Last In First Out)
'''
````

````python
'''
Stack 常見操作：

push：放一個資料進去 stack
pop ：拿出最上面的資料
top ：查看最上面的資料，但不拿出來
is_empty：檢查 stack 是否為空
'''

stack = []

stack.append(10) # push
stack.append(20)
stack.append(30)
stack.append(40)
stack.append(50)
print(stack)

x = stack.pop() # pop
print(x)
print(stack)

print(stack[-1]) # top

print(len(stack) == 0) # is_empty
````

````python
# 用 Class 手刻 Stack

# __init__, push, pop, top, is_empty

class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
          return None
        return self.items.pop()

    def top(self):
        if self.is_empty():
          return None
        return self.items[-1]


s = Stack()

s.push(100)
s.push(200)
s.push(300)

print(s.top())
print(s.pop())
print(s.pop())
````

# 練習 1：反轉字串

請寫一個函式 `reverse_string(s)`，用 Stack 把字串反轉。

範例：

```python
reverse_string('hello')
```

輸出：

```python
'olleh'
```

````python
def reverse_string(s):
    stack = []
    for char in s:
        stack.append(char)

    reverse_done = ""
    while not len(stack) == 0:
        reverse_done += stack.pop()

    return reverse_done

print(reverse_string('hello'))
print(reverse_string('python'))
````

# 練習 2：括號配對

請寫一個函式 `is_valid_parentheses(s)`，判斷括號是否正確配對。

只會出現：`()[]{}`

範例：

```python
is_valid_parentheses('()[]{}') # True
is_valid_parentheses('([{}])') # True
is_valid_parentheses('(]')     # False
is_valid_parentheses('([)]')   # False
```

````python
def is_valid_parentheses(s):
  stack = []
  pair = {
      ')': '(',
      ']': '[',
      '}': '{'
  }

  for ch in s:
    if ch in '([{': # 看到左邊 -> 放進去 stack
      stack.append(ch)
    else: # 看到右邊 -> 1.是不是空 2.從stack拿出，並比較是否一對
      if len(stack) == 0:
        return False

      last = stack.pop()
      if last != pair[ch]:
        return False

    return len(stack) == 0


print(is_valid_parentheses('()[]{}'))
print(is_valid_parentheses('([{}])'))
print(is_valid_parentheses('(]'))
print(is_valid_parentheses('([)]'))
````

````python
# queue 佇列

'''
可以想像成排隊買飲料：先排隊的人，會先被服務

先進先出 FIFO (First In First Out)
'''
````

````python
'''
Queue 常見操作：

enqueue：把資料放到隊伍最後面
dequeue：把資料從隊伍最前面拿出來
front：查看最前面的資料
is_empty：檢查 queue 是否為空
'''

from collections import deque

q = deque()

q.append(10) # enqueue
q.append(20)
q.append(30)

print(q)

x = q.popleft() # dequeue
print(x)
print(q)

print(q[0]) # front

print(len(q) == 0) # is_empty
````
