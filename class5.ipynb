````python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
          return None
        return self.items.pop()

    def top(self):
        if self.is_empty():
          return None
        return self.items[-1]
````

# 練習 3：移除相鄰重複字元

給一個字串，如果有兩個相鄰且相同的字元，就把它們消掉。

用 Stack 解。

範例：

```python
remove_duplicates('abbaca')
```

過程：

```python
abbaca -> aaca -> ca
```

答案：

```python
'ca'
```

````python
def remove_duplicates(s):
  stack = []

  for ch in s:
    # 當 stack 不為空 且 stack 的 top 跟我現在的 ch 是一樣的時候 -> 拿出來
    # 否則 直接放進去
    if len(stack) > 0 and stack[-1] == ch:
      stack.pop()
    else:
      stack.append(ch)

  return ''.join(stack)

print(remove_duplicates('abbaca')) # ca
print(remove_duplicates('azxxzyx')) # ayx
````

````python
# Queue 常見操作

from collections import deque

q = deque()

q.append(10) # enqueue
q.append(20)
q.append(30)

print(q)

x = q.popleft() # dequeue
print(x)
print(q)

print(q[0]) # front
print(len(q) == 0) # is_empty
````

````python
# 為什麼 queue 不用 pop(0) 改用 popleft()

'''
Python list 的 pop(0) 會把第一個元素拿掉
但是拿掉後，後面的每個元素都要往前移動，所以效率比較差
資料少的時候沒差，資料多的時候建議使用 collections.deque
'''
````

````python
# 用 Class 手刻 Queue

from collections import deque

class Queue:
    def __init__(self):
        self.items = deque()

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.items.popleft()

    def front(self):
        if self.is_empty():
            return None
        return self.items[0]

q = Queue()

q.enqueue('A')
q.enqueue('B')
q.enqueue('C')

print(q.front())
print(q.dequeue())
print(q.dequeue())
print(q.is_empty())
````

# 練習 4：模擬排隊

有一群人依序排隊，每次服務最前面的人。

請寫一個函式 `serve_people(names)`，回傳服務順序。

範例：

```python
serve_people(['Amy', 'Bob', 'Cindy'])
```

輸出：

```python
['Amy', 'Bob', 'Cindy']
```

````python
from collections import deque

def serve_people(people):
    input = deque(people)
    result = []
    while input:
        person = input.popleft()
        result.append(person)
    return result

print(serve_people(["Alice", "Bob", "Charlie"]))
````

# 練習 5：最近的請求次數

給一串時間 `times`，每次有一個新的請求時間 `t`。

請回傳每次在最近 3000 毫秒內有幾個請求。

範例：

```python
times = [1, 100, 3001, 3002]
```

輸出：

```python
[1, 2, 3, 3]
```

這題可以用 Queue 保留目前還在範圍內的時間。

````python
from collections import deque

def recent_counter(times):
    input = deque()
    result = []

    for time in times:
        while input and input[0] < time - 3000:
            input.popleft()
        input.append(time)
        result.append(len(input))
    return result


print(recent_counter([1, 100, 3001, 3002])) # [1, 2, 3, 3]
print(recent_counter([1, 2, 3, 4000, 7000])) # [1, 2, 3, 1, 2]
````

````python
# recursion 遞迴

# 函式自己呼叫自己，也就是一個問題太大時，先把它拆成比較小的同類型問題
````

````python
'''
最簡單的遞迴範例：倒數

countdown(3)
會印出：
3
2
1
結束
'''

def countdown(n):
  if n == 0:
    print('結束')
    return

  print(n)
  countdown(n-1)


countdown(3)
````

````python
# 遞迴最重要的兩個部分

'''
1. base case 終止條件
  - 什麼時候要停下來
  - 如果沒有 base case，程式會一直呼叫自己
2. recursive case 遞迴條件
  - 把問題變小
  - 呼叫自己處理比較小的問題
'''
````

````python
def countdown(n):
  # base case
  if n == 0:
    print('結束')
    return

  # recursive case
  print(n)
  countdown(n-1)


countdown(3)
````

````python
'''
如果沒有 Base Case 會怎樣

Python 最後會出現：RecursionError: maximum recursion depth exceeded

這代表遞迴太多層，程式停不下來
'''
````

# Recursion 和 Stack 的關係

前面學過 Stack：

**後進先出 LIFO**

其實每次函式被呼叫時，電腦都會把它放進一個叫做 **Call Stack** 的地方

所以 recursion 本質上和 stack 很有關係

````python
'''
觀察遞迴的進入與離開

會先一路進去：
進入 3
進入 2
進入 1
進入 0

遇到 base case 後，再一層一層回來：
離開 1
離開 2
離開 3
'''

def demo(n):
  print('進入', n)

  if n == 0:
    print('遇到 base case')
    return

  demo(n - 1)

  print('離開', n)

demo(3)
````

# 範例 1：階乘 Factorial

階乘寫法：

```python
5! = 5 * 4 * 3 * 2 * 1
```

所以：

```python
factorial(5) = 120
```

````python
def factarial(n):
    if n == 0:
        return 1
    else:
        return n * factarial(n - 1)

print(factarial(5))
````

# 階乘的遞迴想法

```python
factorial(5)
= 5 * factorial(4)
= 5 * 4 * factorial(3)
= 5 * 4 * 3 * factorial(2)
= 5 * 4 * 3 * 2 * factorial(1)
= 5 * 4 * 3 * 2 * 1 * factorial(0)
= 120
```

這裡的 base case 是：

```python
factorial(0) = 1
```

# 範例 2：費氏數列 Fibonacci

Fibonacci 定義：

```python
fib(0) = 0
fib(1) = 1
fib(n) = fib(n - 1) + fib(n - 2)
```

數列長這樣：

```python
0, 1, 1, 2, 3, 5, 8, 13, ...
```

````python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)


for i in range(8):
  print(fibonacci(i))
````

# 範例 3：字串反轉

問題：

```python
reverse('hello')
```

答案：

```python
'olleh'
```

````python
def reverse(s):
  if len(s) == 0:
    return ''

  return reverse(s[1:]) + s[0]

print(reverse('hello'))
print(reverse('python'))
````
