````python
'''
DFS (Depth First Search) 深度優先搜尋
1. 定義問題：從一個起點開始，沿著可以前往的方向，不斷找下去
2. graph, visited
3. 遞迴流程：
  拜訪目前節點
  → 標記為走過
  → 找出所有相鄰節點
  → 對尚未走過的相鄰節點繼續 DFS
4. DFS 實作會使用 stack
'''
````

````python
# https://leetcode.com/problems/pascals-triangle/description

def generate(numRows):
    result = []
    for i in range(numRows):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = result[i - 1][j - 1] + result[i - 1][j]
        result.append(row)
    return result

print(generate(5))
````

# BFS：廣度優先搜尋

BFS 全名是：

> Breadth-First Search

中文稱為：

> 廣度優先搜尋

BFS 的搜尋方式是：

> 先搜尋距離起點最近的位置，再逐層向外擴散。

可以想像水波：

```text
第 0 層：起點
第 1 層：與起點相鄰的位置
第 2 層：距離起點兩步的位置
第 3 層：距離起點三步的位置
```

BFS 使用 Queue，因為先加入 Queue 的節點要先處理。

````python
'''
A ─ B ─ D
│   │
C   E
'''

from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A"],
    "D": ["B"],
    "E": ["B"]
}

queue = deque(["A"])
visited = {"A"}

while queue: # A -> B, C -> C, D, E -> D, E -> E -> []
    node = queue.popleft()
    print("走到：", node)

    for next_node in graph[node]:
        if next_node not in visited:
            visited.add(next_node)
            queue.append(next_node)
````

# BFS Step by Step

從 A 開始：

```text
A ─ B ─ D
│   │
C   E
```

Queue 的變化：

|步驟|取出|新加入|Queue|
|---|---|---|---|
|開始|-|A|A|
|1|A|B、C|B、C|
|2|B|D、E|C、D、E|
|3|C|無|D、E|
|4|D|無|E|
|5|E|無|空|

走訪順序：

```text
A → B → C → D → E
```

BFS 會先處理與 A 距離 1 步的 B、C，再處理距離 2 步的 D、E。

````python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A"],
    "D": ["B"],
    "E": ["B"]
}

def bfs(graph ,start):
    visited = set()
    queue = [start]
    visited.add(start)

    while queue:
        node = queue.pop(0)
        print('走到' + node)

        for a in graph[node]:
            if a not in visited:
                visited.add(a)
                queue.append(a)

bfs(graph, 'A')
````

# DFS 和 BFS 比較

|項目|DFS|BFS|
|---|---|---|
|中文|深度優先搜尋|廣度優先搜尋|
|主要工具|Recursion / Stack|Queue|
|搜尋方式|一條路走到底|一層一層擴散|

簡單記法：

```text
DFS：走到底再回來
BFS：附近找完再往外
```

````python
# 樹 Tree

'''
重要術語：
1. 根節點 root：樹的最頂端節點（沒有父節點）
2. 父節點 parent / 子節點 child：相連的上下級關係
3. 葉子節點 leaf：樹的最末端節點（沒有子節點）
4. 深度 depth / 高度 height：從 Root 到該節點經過的邊數或層數
5. 環 cycle：「樹裡面沒有環」，從 Root 到任意節點只有唯一一條路徑，因此搜尋樹時不需要 visited
'''

# 二元樹 Binary Tree: 每個節點最多只能有兩個子節點，分別稱為左子節點 (Left Child) 與 右子節點 (Right Child)
````

````python
# 表示二元樹

# 方法 A：使用類別 (Class / 指標表示法)
class TreeNode:
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

'''
     1
    / \
   2   3
  / \
 4   5
'''

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

print("根節點的值：", root.val)
print("左子節點的值：", root.left.val)
print("左子節點的右子節點：", root.left.right.val)
````

````python
# 方法 B：使用陣列 (List 陣列表示法)

# 完全二元樹 (Complete Binary Tree)，可用 python list 來存

'''
規則（假設 Root 在索引 0）：

索引 i 的左子節點在：2 * i + 1
索引 i 的右子節點在：2 * i + 2
索引 i 的父節點在：(i - 1) // 2
'''

tree_list = ["A", "B", "C", "D", "E", "F", "G"]
#         A(0)
#       /      \
#     B(1)     C(2)
#    /   \    /   \
#  D(3) E(4) F(5) G(6)

idx = 2 # 節點 C
left_child = 2 * idx + 1
right_child = 2 * idx + 2

print(f"節點 {tree_list[idx]} 的左小孩是: {tree_list[left_child]}")
print(f"節點 {tree_list[idx]} 的右小孩是: {tree_list[right_child]}")
````
