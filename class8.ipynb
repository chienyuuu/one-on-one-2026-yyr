````python
'''
BFS 廣度優先搜尋
1. 先搜尋距離起點最近的位置，再「逐層」向外擴散
2. 常使用的資料結構：Queue
'''
````

````python
'''
A ─ B ─ D
│   │
C   E
'''

from collections import deque

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A"],
    "D": ["B"],
    "E": ["B"]
}

queue = deque(['A'])
visited = {'A'}

while queue:
  node = queue.popleft()
  print('走到', node)

  for next_node in graph[node]:
    if next_node not in visited:
      visited.add(next_node)
      queue.append(next_node)
````

````python
'''
Tree 樹

重要術語
  - 根節點 root：樹的最頂端節點（沒有父節點）
  - 父節點 parent / 子節點 child：相連的上下級關係
  - 葉子節點 leaf：樹的最末端節點（沒有子節點）
  - 深度 depth / 高度 height：從 Root 到該節點經過的邊數或層數
  - 環 cycle：樹裡面沒有環，從 Root 到任意節點只有唯一一條路徑，因此搜尋樹時不需要 visited


Binary Tree 二元樹

1. 每個節點最多只能有兩個子節點，分別稱為左子節點 (Left Child) 與 右子節點 (Right Child)
2. 表示二元樹
  - 方法 A：使用類別 (Class 指標表示法)
  - 方法 B：使用串列 (List 串列表示法)
'''
````

````python
# 方法 A：使用類別 (Class / 指標表示法)

'''
     1
    / \
   2   3
  / \
 4   5
'''

class TreeNode:
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

print("根節點的值：", root.val)
print("左子節點的值：", root.left.val)
print("左子節點的右子節點：", root.left.right.val)
````

````python
# 方法 B：使用串列 (List 串列表示法)

'''
完全二元樹 (Complete Binary Tree)
1. 有右子就必須有左子，但是有左子不一定需要右子
2. 由上到下、由左到右長滿節點
'''


'''
規則（假設 Root 在索引 0）：

索引 i 的左子節點在：2 * i + 1
索引 i 的右子節點在：2 * i + 2
索引 i 的父節點在：(i - 1) // 2
'''

tree_list = ["A", "B", "C", "D", "E", "F", "G"]
#         A(0)
#       /      \
#     B(1)     C(2)
#    /   \    /   \
#  D(3) E(4) F(5) G(6)

idx = 2 # 節點 C
left_child = 2 * idx + 1
right_child = 2 * idx + 2

print(f"節點 {tree_list[idx]} 的左小孩是: {tree_list[left_child]}")
print(f"節點 {tree_list[idx]} 的右小孩是: {tree_list[right_child]}")
````

# 練習 1：計算葉子節點數量 (Count Leaves)

請寫一個遞迴函式 `count_leaves(node)`，計算二元樹中有幾張「葉子」（即沒有左右小孩的節點）。

提示：
1. 如果 `node` 是 `None`，回傳 `0`
2. 如果 `node.left` 和 `node.right` 都是 `None`，代表它是葉子，回傳 `1`
3. 否則回傳 `左子樹葉子數 + 右子樹葉子數`

````python
class TreeNode:
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

def count_leaves(node):
  if node is None:
    return 0
  if node.left is None and node.right is None:
    return 1
  return count_leaves(node.left) + count_leaves(node.right)

'''
     10
    /  \
  5    15
  /       \
3         20
'''

root = TreeNode(10)
root.left = TreeNode(5)
root.right = TreeNode(15)
root.left.left = TreeNode(3)
root.right.right = TreeNode(20)
root.right.left = TreeNode(100)

print('葉子數量', count_leaves(root))
````

````python
# 二元搜尋樹 Binary Search Tree (BST)

'''
1. 左子樹中所有節點的值都小於「當前節點」的值
2. 右子樹中所有節點的值都大於「當前節點」的值
3. 左右子樹也必須各自是一棵 BST
'''

'''
       8 (Root)
      / \
     3   10
    / \    \
   1   6    14
'''

# BST 好處：搜尋某個數值的時候，每次比較可排除一半的節點，平均搜尋的時間複雜度 O(log N)
````

````python
class TreeNode:
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

def search_bst(root, target):
  if not root or root.val == target:
    return root

  if target < root.val:
    return search_bst(root.left, target)
  else:
    return search_bst(root.right, target)

# 建 BST
bst_root = TreeNode(8)
bst_root.left = TreeNode(3)
bst_root.right = TreeNode(10)
bst_root.left.left = TreeNode(1)
bst_root.left.right = TreeNode(6)
bst_root.right.right = TreeNode(14)

target_node = search_bst(bst_root, 6)
print("搜尋 6 的結果：", target_node.val if target_node else "未找到")
target_node = search_bst(bst_root, 9)
print("搜尋 9 的結果：", target_node.val if target_node else "未找到")
````

# 樹的 DFS 三種走訪（Pre, In, Post-order）

深度優先走訪（DFS）依據「根節點拜訪的時機」分為三種：

1. **前序走訪 (Pre-order Traversal)**：`根` → `左` → `右`
2. **中序走訪 (In-order Traversal)**：`左` → `根` → `右`（**極度重要：BST 的中序走訪結果會是嚴格遞增排序！**）
3. **後序走訪 (Post-order Traversal)**：`左` → `右` → `根`（常用於刪除樹、計算子樹大小）

對應範例樹：
```text
     1
    / \
   2   3
  / \
 4   5
```

- **前序 (Pre-order)**：`1 -> 2 -> 4 -> 5 -> 3`
- **中序 (In-order)**：`4 -> 2 -> 5 -> 1 -> 3`
- **後序 (Post-order)**：`4 -> 5 -> 2 -> 3 -> 1`

````python
# 走訪方法：https://i.sstatic.net/LY89o.png
````

````python
class TreeNode:
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

# 用遞迴實作 DFS 的三種走訪

def preorder(node):
    if not node:
        return []
    return [node.val] + preorder(node.left) + preorder(node.right)

def inorder(node):
    if not node:
        return []
    return inorder(node.left) + [node.val] + inorder(node.right)

def postorder(node):
    if not node:
        return []
    return postorder(node.left) + postorder(node.right) + [node.val]

print("遞迴 Pre-order :", preorder(root))
print("遞迴 In-order  :", inorder(root))
print("遞迴 Post-order:", postorder(root))
````

# 練習 3：驗證是否為合法 BST (Validate BST)

給定一顆二元樹，請檢查它是不是合法的 BST。

**解法**：做一次 **In-order Traversal**，檢查印出來的串列是不是**嚴格遞增**！

````python
'''
1 2 3 3 3 3 3 4 5 6 7 遞增
1 2 3 4 5 6 7 嚴格遞增
'''
def is_valid_bst(root):
  vals = inorder(root)
  # 檢查嚴格遞增
  for i in range(len(vals) - 1):
    if vals[i] >= vals[i+1]:
      return False
  return True

# 測資一
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
print(inorder(root))
print("root 是合法 BST 嗎？", is_valid_bst(root))

# 測資二
bst_root = TreeNode(8)
bst_root.left = TreeNode(3)
bst_root.right = TreeNode(10)
bst_root.left.left = TreeNode(1)
bst_root.left.right = TreeNode(6)
bst_root.right.right = TreeNode(14)
print(inorder(bst_root))
print("bst_root 是合法 BST 嗎？", is_valid_bst(bst_root))
````
