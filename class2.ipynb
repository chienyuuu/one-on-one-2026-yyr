```python
'''
字典 dictionary
1. 鍵 key - 值 value
2. dict1 = {'John': 30, 'Bob': 20} (string, integer, float 都可以當 key，只是大部分的鍵都是 string)
3. 當鍵重複時，後面會覆蓋掉前面，只有最後的有效
4. 當鍵不存在時，直接 print(dict['Amy']) 會出錯 -> 使用 get()
5. get(鍵, 預設值)
  - 鍵存在 -> 回傳值
  - 鍵不存在 且 沒有預設值 -> 回傳 None
  - 鍵不存在 但 有預設值 -> 回傳 預設值
6. 修改/新增字典 dict1['John'] = 31
  - 若 key 存在 -> 修改/取代
  - 若 key 不存在 -> 新增
7. 刪除字典
  - 刪除字典中特定元素 del dict1['Amy']
  - 刪除字典所有元素 dict1.clear()
  - 刪除整個字典 del dict1
8. 字典個數 len()
9. 字典複製 dict2 = dict1.copy()
10. 判斷 key 是否存在: 鍵 in 字典名稱
11. keys()：可取得所有鍵，資料型態是 dict_keys
12. values()：可取得所有值，資料型態是 dict_values
13. items()：以「鍵值對」取得元素組合，資料型態是 dict_items
14. setdefault(鍵, 預設值)：和 get() 類似，若 key 不存在，就用括號內的參數「建立新元素」
'''
```

```python
'''
range()
1. 一個參數：數列變數 = range(整數值) -> 產生 0 至 (整數值 - 1) 的串列
2. 兩個參數：數列變數 = range(起始值, 終止值) -> 產生 (起始值) 至 (終止值 - 1) 的串列
3. 三個參數：數列變數 = range(起始值, 終止值, 間隔值) -> 產生 (起始值) 開始，每次遞增 (間隔值)，直到 (終止值 - 1) 的串列
'''
```

```python
'''
zerojudge: https://zerojudge.tw/ShowProblem?problemid=a058$0
'''
n = int(input())
count = {0 : 0, 1 : 0, 2 : 0}

for _ in range(n):
    num = int(input())

    r = num % 3
    count[r] += 1

print(count[0], count[1], count[2])
```

```python
# 物件 object

'''
可以想像成一個人，每個人都有自己的屬性，例如：年齡、性別、身高

如果我們要使用一個人的物件，我們可以像這樣使用它：

print(People.age)
print(People.height)
'''
```

```python
# 類別 class

# 類別就是定義一個物件需要有什麼樣的「屬性」或是「函式」

class Person: # 定義 Person 類別
  def __init__(self, age, height): # 初始化函式
    self.age = age
    self.height = height

  def birthday(self): # 其他自定義函式
    self.age += 1



amy = Person(18, 160)

print(amy.age)

amy.birthday()

print(amy.age)
```

```python
'''
(1) class 類別名稱 : 類別關鍵字
類通常自訂類別使用大寫作為開頭，例如：Person

(2) init(self): 初始化函式
初始化函式用來定義該類別的初始屬性，例如：age, height
其中的 self 代表當下的物件本身，所以可以看成“自己的age“

(3)自定義函式：
使用自定義函式時，可以將物件本身進行改動，例如：改變屬性等
若要使用物件屬性，需要將“self”當作參數，使用self.屬性即可讀取
'''
```

```python
class Car:
  def __init__(self, brand, price):
    self.brand = brand
    self.price = price

  def discount(self, discount):
    self.price *= discount

  def __str__(self):
    return f"{self.brand} - {self.price}"

car1 = Car('benz', 300)
car2 = Car('toyota', 150)

print(car1)
print(car2)

car1.discount(0.8)

print(car1)
print(car2)
```

```python
'''
練習

請設計一個 Player 類別，功能如下：
建立物件時需要輸入玩家名字 name，血量 hp 預設為 100
實作 attack() 方法，使血量減少 10
實作 heal() 方法，使血量增加 5，但最多不能超過 100
實作 show() 方法，印出玩家名字與目前血量
'''
class Player:
    def __init__(self, name):
        self.name = name
        self.hp = 100
    def attack(self):
        self.hp -= 10
    def heal(self):
        if self.hp < 100:
            self.hp += 5
    def show(self):
        print(f'{self.name} 的hp為 {self.hp}')

'''
輸入

p1 = Player("Alex")

p1.attack()
p1.attack()
p1.heal()

p1.show()
'''

'''
輸出

玩家: Alex, HP: 85
'''
```

```python
# 字串格式化 %
# print(項目 % (參數列))

'''
常用參數
%d -> 整數
%s -> 字串
%f -> 浮點數
%% -> %
%e / %E -> 科學記號
'''

print("" % ())
print("%s 的成績是 %d" % ('amy', 90))
print("小明成績是班上 %d %%" % (90))
```

```python
# 精確控制輸出位置 -> 排列整齊

# %5d：固定列印5個字元，少於5個會在左邊補空白字元
print("%5d" % (123))
print("%5d" % (12345))
print("%5d" % (1234567))
print("%5s" % ('嗨嗨'))
```

```python
# %8.2f：固定8個字元（包含小數點），小數點後固定列印2個字元
# 整數不足 -> 左邊補空白字元
# 小數不足 -> 右邊補 0
print("%8.2f" % (123.4567))
print("%8.2f" % (123.4))
```

```python
'''
練習：
國稅局的資料，整齊列出這三年的各種稅收
"2017", 98.34, 90.2, 104.79
"2016", 83, 110.5, 82.45
"2015", 98, 79.32, 102
'''

'''
年度   所得稅   營業稅   證交稅
2017  98.34   90.20  104.79
2016  83.00  110.50   82.45
2015  98.00   79.32  102.00
'''
print("年度  所得稅  營業稅  證交稅")
print("%4s %6.2f %6.2f %6.2f" % ("2017", 98.34, 90.2, 104.79))
print("%4s %6.2f %6.2f %6.2f" % ("2016", 83, 110.5, 82.45))
print("%4s %6.2f %6.2f %6.2f" % ("2015", 98, 79.32, 102))
```

```python
# 字串格式化 format
# 用大括號 {} 表示參數位置就好
# print(字串.format(參數列))

print("".format())
print('{} 的成績是 {} 分'.format('小明', 80))

# 多個參數，可在大括號放順序數字（從 0 開始）
print('{0} 的成績是 {1} 分'.format('小明', 80))
print('考 {1} 的人是 {0}'.format('小明', 80))
```

```python
# 字串不足 -> 空白補右邊
# 數字不足 -> 空白補左邊

print('{}的成績是{}分'.format('amy', 80))
print('{0}的成績是{1}分'.format('amy', 80))

print('{0:7s}的成績是{1:3d}分'.format('amy', 80))
print('{0:7}的成績是{1:3}分'.format('amy', 80))
print('{:7}的成績是{:3}分'.format('amy', 80))
```
